# Proof That the Type-(iv) Exclusion in the CMV Strip Problem Extends to $\lambda\ge 1.2581840884$

Let

$$
f_\lambda(x,y)=\begin{cases}1,&|y|\le1,\\ \lambda,&|y|>1,\end{cases}
\qquad \lambda>1.
$$

For a regular planar region $E$, write

$$
A_\lambda(E):=\int_E f_\lambda\,d\mathcal L^2,
\qquad
P_\lambda(E):=\int_{\partial E}f_\lambda\,d\mathcal H^1.
$$

### Theorem

Let $E_{\mathrm{iv}}$ be a regular type-(iv) candidate in the sense of Cañete–Miranda Jr.–Vittone. If $\lambda\ge1.2581840884$, then there is a regular competitor $E_{\mathrm{new}}$ such that

$$
A_\lambda(E_{\mathrm{new}})=A_\lambda(E_{\mathrm{iv}}),
\qquad
P_\lambda(E_{\mathrm{new}})<P_\lambda(E_{\mathrm{iv}}).
$$

Hence $E_{\mathrm{iv}}$ is not isoperimetric. A type-(iv) minimizer can occur only if

$$\boxed{1<\lambda<1.2581840884}.$$


## CMV Results Used as Prerequisites

The following facts are cited and are not reproved here.

1. Planar isoperimetric regions for the strip density exist and are connected. **[CMV, Proposition 3.5]**
2. Schwarz symmetrization gives an isoperimetric representative whose horizontal slices are intervals; a region meeting one exterior component must meet both. **[CMV, Proposition 3.6]**
3. The vertically symmetric candidates are precisely families (i)–(iv). A type-(iv) candidate is bounded by four equal-radius circular arcs: two in the strip and two reflected exterior arcs satisfying Snell's law. **[CMV, Lemma 3.8]**
4. Every isoperimetric region has vertical reflective symmetry. **[CMV, Proposition 3.9]**
5. Type-(iv) candidates have curvature $h$ and angles
   $$
   \widehat\beta=\frac\pi2-\arcsin h,
   \qquad
   \widehat\alpha=\arccos\!\left(\frac{\cos\widehat\beta}{\lambda}\right),
   \qquad0<h\le1.
   $$
   Since $\cos\widehat\beta=h$, this is equivalent to $\lambda\cos\widehat\alpha=h$. **[CMV, equation (27)]**
6. If either exterior cap has chord $L$ and Euclidean area $A$, replacing the two congruent exterior caps by one cap of area $2A$ and one exposed chord reduces perimeter exactly when
   $$
   2\operatorname{arc}(x)-\operatorname{arc}(2x)>\frac1\lambda,
   \qquad x:=\frac{A}{L^2}.
   $$
   CMV prove the lower bound $2\operatorname{arc}(x)-\operatorname{arc}(2x)>\pi/4$, obtaining exclusion for $\lambda\ge4/\pi$. **[CMV, proof of Theorem 3.16]**
7. CMV denote the true minimum by $k$ and state that the same argument holds for $\lambda>1/k$, but do not compute $k$. **[CMV, Remark 3.17]**

Only the computation of this minimum, its rational certification, and the explicit regular-frontier realization used below are new.


## 1. Circular segments with unit chord

Let $0<\theta<\pi$ be the half-central-angle of a circular segment with chord $1$. Its radius, arc length, and enclosed area are

$$
r(\theta)=\frac1{2\sin\theta},
\qquad
\ell(\theta)=2r(\theta)\theta=\frac{\theta}{\sin\theta},
\tag{1}
$$

$$
a(\theta)=r(\theta)^2(\theta-\sin\theta\cos\theta)
=\frac{\theta-\sin\theta\cos\theta}{4\sin^2\theta}.
\tag{2}
$$

Differentiation gives

$$
\ell'(\theta)=\frac{\sin\theta-\theta\cos\theta}{\sin^2\theta},
\qquad
a'(\theta)=\frac{\sin\theta-\theta\cos\theta}{2\sin^3\theta}.
\tag{3}
$$

For $q(\theta):=\sin\theta-\theta\cos\theta$,

$$q(0)=0,
\qquad q'(\theta)=\theta\sin\theta>0\quad(0<\theta<\pi).$$

Thus $q(\theta)>0$ and $a'(\theta)>0$. Moreover,

$$\lim_{\theta\to0^+}a(\theta)=0,
\qquad\lim_{\theta\to\pi^-}a(\theta)=+\infty.$$

Hence $a:(0,\pi)\to(0,\infty)$ is bijective. Define

$$
\theta(x):=a^{-1}(x),
\qquad
\operatorname{arc}(x):=\ell(\theta(x)),
\qquad x>0.
\tag{4}
$$

The interval $0<\theta<\pi$ includes minor, semicircular, and major arcs.

*Lean formalization:* [`MorgansArcFunction.lean`](MorgansArcFunction.lean), `ell`, `area`, `θOf`, `arc`, `area_strictMonoOn`, `area_surjOn`, `area_θOf`, `θOf_area`.


## 2. Derivative of the arc function

The inverse-function theorem and chain rule yield

$$
\operatorname{arc}'(x)
=\frac{\ell'(\theta(x))}{a'(\theta(x))}.
$$

Using (3), the common factor $q(\theta)$ cancels:

$$
\begin{aligned}
\operatorname{arc}'(x)
&=\frac{(\sin\theta-\theta\cos\theta)/\sin^2\theta}
{(\sin\theta-\theta\cos\theta)/(2\sin^3\theta)}\\
&=2\sin\theta,
\end{aligned}
$$

where $\theta=\theta(x)$. Therefore

$$\boxed{\operatorname{arc}'(x)=2\sin\theta(x)}.\tag{5}$$

This exact identity replaces the qualitative convexity/concavity estimate used in CMV.

*Lean formalization:* [`MorgansArcFunction.lean`](MorgansArcFunction.lean), `ell_div_area_deriv`, `hasDerivAt_arc`.


## 3. Derivative of the CMV comparison function

Define

$$g(x):=2\operatorname{arc}(x)-\operatorname{arc}(2x),\qquad x>0.\tag{6}$$

This is the function denoted $h$ in the proof of CMV Theorem 3.16. Equation (5) gives

$$
\boxed{g'(x)=4\bigl(\sin\theta(x)-\sin\theta(2x)\bigr)}.
\tag{7}
$$

A unit-chord semicircle has area $\pi/8$, so

$$a(\pi/2)=\pi/8,
\qquad\theta(\pi/8)=\pi/2.\tag{8}$$

If $0<x\le\pi/16$, then

$$0<\theta(x)<\theta(2x)\le\pi/2,$$

so strict increase of sine gives $g'(x)<0$. If $x\ge\pi/8$, then

$$\pi/2\le\theta(x)<\theta(2x)<\pi,$$

so strict decrease of sine gives $g'(x)>0$. Every minimizer therefore lies in

$$\boxed{\frac\pi{16}<x<\frac\pi8}.\tag{9}$$

## 4. Critical-point equation

Let $x$ satisfy (9) and $g'(x)=0$. Then

$$\sin\theta(x)=\sin\theta(2x).\tag{10}$$

Since

$$0<\theta(x)<\frac\pi2<\theta(2x)<\pi,$$

the angles in (10) are supplementary:

$$\theta(2x)=\pi-\theta(x).\tag{11}$$

Writing $\theta=\theta(x)$, equation (11) is equivalent to

$$a(\pi-\theta)=2a(\theta).$$

Using (2),

$$
\frac{\pi-\theta+\sin\theta\cos\theta}{4\sin^2\theta}
=2\frac{\theta-\sin\theta\cos\theta}{4\sin^2\theta},
$$

which reduces to

$$\boxed{3\theta-3\sin\theta\cos\theta=\pi}.\tag{12}$$

Define

$$F(\theta):=3\theta-3\sin\theta\cos\theta-\pi.$$

Then

$$F'(\theta)=6\sin^2\theta>0\qquad(0<\theta<\pi),$$

and

$$F(0)=-\pi<0,
\qquad F(\pi/2)=\pi/2>0.$$

There is therefore a unique $\theta_*\in(0,\pi/2)$ satisfying $F(\theta_*)=0$.

*Lean formalization:* [`Minimiser.lean`](Minimiser.lean), `F`, `hasDerivAt_F`, `F_strictMonoOn`, `θstar_spec`, `θstar_unique`, `area_pi_sub_eq_two_mul`.


## 5. Exact global minimum

Set $x_*:=a(\theta_*)$. Equation (12) gives

$$a(\pi-\theta_*)=2a(\theta_*),$$

so bijectivity of $a$ implies

$$\theta(2x_*)=\pi-\theta_*.$$

Using (1),

$$
\begin{aligned}
g(x_*)
&=2\ell(\theta_*)-\ell(\pi-\theta_*)\\
&=2\frac{\theta_*}{\sin\theta_*}
 -\frac{\pi-\theta_*}{\sin\theta_*}\\
&=\frac{3\theta_*-\pi}{\sin\theta_*}.
\end{aligned}
$$

Since $F(\theta_*)=0$,

$$3\theta_*-\pi=3\sin\theta_*\cos\theta_*,$$

and hence

$$\boxed{g(x_*)=3\cos\theta_*}.\tag{13}$$

To prove global minimality, define

$$d(x):=\sin\theta(x)-\sin\theta(2x).$$

Because $\theta(2x_*)=\pi-\theta_*$, equation (7) gives $g'(x_*)=0$. The derivative signs in Section 3 therefore force $\pi/16<x_*<\pi/8$. On this interval, $\sin\theta(x)$ strictly increases while $\sin\theta(2x)$ strictly decreases; hence $d$ is strictly increasing. Since $g'(x)=4d(x)$ and $g'(x_*)=0$, the derivative is negative before $x_*$ and positive after $x_*$. The signs from Section 3 cover the remainder of $(0,\infty)$. Thus

$$g'(x)<0\quad(0<x<x_*),
\qquad g'(x)>0\quad(x>x_*),$$

and

$$
\boxed{\min_{x>0}g(x)=3\cos\theta_*},
\qquad
\boxed{3\cos\theta_*\le g(x)\quad(x>0)}.
\tag{14}
$$

Numerically,

$$
\theta_*=1.302662837300451\ldots,
\qquad x_*=0.281561941577875\ldots,
$$

$$3\cos\theta_*=0.794796253808331\ldots.$$

## 6. Exact rational certificate

The infimum threshold for uniform strict improvement by this cap substitution is

$$
\lambda_*:=\frac1{3\cos\theta_*}
=1.258184088322534\ldots.
\tag{15}
$$

Uniform strict improvement holds for $\lambda>\lambda_*$. At $\lambda=\lambda_*$, the perimeter difference is zero for $x=x_*$.

The formal exclusion uses the rational number

$$\lambda_0:=1.2581840884>\lambda_*.$$

Let $t_0:=1.30266283731$. The exact rational bounds

$$
3.14159265358979323846<\pi
<3.14159265358979323847,
\tag{16}
$$

$$\sin(2t_0)\le0.510930572191682027065\tag{17}$$

give

$$
\begin{aligned}
F(t_0)
&=3t_0-\pi-\frac32\sin(2t_0)\\
&>3t_0-3.14159265358979323847
 -\frac32(0.510930572191682027065)\\
&=5.26837209325\times10^{-11}>0.
\end{aligned}
\tag{18}
$$

Because $F$ is strictly increasing and $F(\theta_*)=0$, one obtains $\theta_*<t_0$. A second exact interval calculation yields

$$
0.264932084590720839353931
\le\cos t_0
\le0.264932084602032613060016.
\tag{19}
$$

The bounds (17) and (19) use small-angle rational enclosures and the exact recurrences

$$
\sin(2u)=2\sin u\cos u,
\qquad
\cos(2u)=1-2\sin^2u.
\tag{20}
$$

For (17), the initial angle is $(\pi-2t_0)/256$, followed by eight doublings and $\sin(2t_0)=\sin(\pi-2t_0)$. For (19), the initial angle is $t_0/512$, followed by nine doublings. Every interval endpoint is rational. The initial certified enclosures are

$$
\begin{aligned}
u&:=\frac{\pi-2t_0}{256},\\
0.00209479288660075483&<u<0.00209479288660075484,\\
0.002094791354553993674
&\le\sin u
\le0.002094791354554397057,\\
0.004189573516827420428
&\le\sin(2u)
\le0.004189573516840328335,
\end{aligned}
\tag{20a}
$$

and

$$
\begin{aligned}
0.002544260609167936018368
&\le\sin(t_0/512)
\le0.002544260609169002148735,\\
0.005088504748616925646946
&\le\sin(t_0/256)
\le0.005088504748651041818669.
\end{aligned}
\tag{20b}
$$

The remaining bounds follow by repeated application of (20).

Since cosine is strictly decreasing on $(0,\pi)$,

$$
3\cos\theta_*>3\cos t_0
\ge0.794796253772162518061793.
\tag{21}
$$

Moreover,

$$
\frac1{1.2581840884}
=0.794796253759395420438858\ldots
<0.794796253772162518061793,
\tag{22}
$$

with exact positive margin greater than $1.2767\times10^{-11}$. Combining (14), (21), and (22),

$$\boxed{\frac1{1.2581840884}<g(x)\quad(x>0)}.\tag{23}$$

If $\lambda\ge1.2581840884$, reciprocal monotonicity gives

$$\boxed{\frac1\lambda<g(x)\quad(x>0)}.\tag{24}$$

## 7. Regular geometric realization of the CMV substitution

Let the exterior half-angle be $\alpha\in(0,\pi/2)$ and let the common curvature satisfy $0<h\le1$. The radius is $R=1/h$. Either original exterior cap has chord length

$$L=2R\sin\alpha=\frac{2\sin\alpha}{h}>0\tag{25}$$

and Euclidean area

$$
A=R^2(\alpha-\sin\alpha\cos\alpha)
=\frac{\alpha-\sin\alpha\cos\alpha}{h^2}>0.
\tag{26}
$$

Consequently

$$
\frac{A}{L^2}
=\frac{\alpha-\sin\alpha\cos\alpha}{4\sin^2\alpha}
=a(\alpha).
\tag{27}
$$

Set $x=A/L^2$. The original cap arc has length $L\operatorname{arc}(x)$. Define

$$\alpha_{\mathrm{new}}:=\theta(2x).$$

Then $a(\alpha_{\mathrm{new}})=2x$, so the replacement cap has area $2A$ and arc length $L\operatorname{arc}(2x)$.

Let $C$ be the unchanged strip core, with side arcs $\Gamma_{\mathrm L}$ and $\Gamma_{\mathrm R}$. Let $\Gamma_+,\Gamma_-$ be the original exterior arcs, $\Gamma_{2A}$ the replacement arc, and $S_-$ the exposed lower chord. The constructed carriers satisfy

$$
\partial E_{\mathrm{iv}}
=\Gamma_{\mathrm L}\cup\Gamma_{\mathrm R}
 \cup\Gamma_+\cup\Gamma_-,
\tag{28}
$$

$$
\partial E_{\mathrm{new}}
=\Gamma_{\mathrm L}\cup\Gamma_{\mathrm R}
 \cup\Gamma_{2A}\cup S_-.
\tag{29}
$$

The old attachment chords are internal; the upper replacement chord is internal; the lower chord is exactly $S_-$. Components meet only at endpoints, which have zero $\mathcal H^1$ measure. Their lengths are

$$
\mathcal H^1(\Gamma_+)=\mathcal H^1(\Gamma_-)
=L\operatorname{arc}(x),
$$

$$
\mathcal H^1(\Gamma_{2A})=L\operatorname{arc}(2x),
\qquad\mathcal H^1(S_-)=L.
\tag{30}
$$

The construction includes minor, semicircular, and major replacement arcs.

## 8. Equal area and strict perimeter improvement

The original and replacement weighted areas are

$$A_\lambda(E_{\mathrm{iv}})=A_\lambda(C)+2\lambda A,$$

$$A_\lambda(E_{\mathrm{new}})=A_\lambda(C)+\lambda(2A).$$

Therefore

$$\boxed{A_\lambda(E_{\mathrm{new}})=A_\lambda(E_{\mathrm{iv}})}.\tag{31}$$

Let $P_\lambda(C)$ denote the contribution of the unchanged side arcs. Equations (28)–(30) give

$$
P_\lambda(E_{\mathrm{iv}})
=P_\lambda(C)+2\lambda L\operatorname{arc}(x),
\tag{32}
$$

$$
P_\lambda(E_{\mathrm{new}})
=P_\lambda(C)+\lambda L\operatorname{arc}(2x)+L.
\tag{33}
$$

Subtracting and using (6),

$$
\begin{aligned}
P_\lambda(E_{\mathrm{new}})-P_\lambda(E_{\mathrm{iv}})
&=L+\lambda L\operatorname{arc}(2x)-2\lambda L\operatorname{arc}(x)\\
&=L\left(1-\lambda\bigl(2\operatorname{arc}(x)-\operatorname{arc}(2x)\bigr)\right)\\
&=\boxed{L\bigl(1-\lambda g(x)\bigr)}.
\end{aligned}
\tag{34}
$$

For $\lambda\ge1.2581840884$, equation (24) gives $1<\lambda g(x)$, hence $1-\lambda g(x)<0$. Since $L>0$,

$$\boxed{P_\lambda(E_{\mathrm{new}})<P_\lambda(E_{\mathrm{iv}})}.\tag{35}$$

*Lean formalization:* [`GeometricBridge.lean`](GeometricBridge.lean), `cap_replacement_preserves_area`, `cap_replacement_perimeter_difference`, `cap_replacement_strictly_improves`.


## 9. Exclusion of type (iv)

Assume $E_{\mathrm{iv}}$ minimizes weighted perimeter among regions of the same weighted area. By (31), $E_{\mathrm{new}}$ is an admissible equal-area competitor. Minimality implies

$$P_\lambda(E_{\mathrm{iv}})\le P_\lambda(E_{\mathrm{new}}).\tag{36}$$

Equation (35) gives the strict reverse inequality, a contradiction. Therefore no regular type-(iv) candidate minimizes weighted perimeter when

$$\lambda\ge1.2581840884.$$

Since the strip problem assumes $\lambda>1$, any type-(iv) minimizer must satisfy

$$\boxed{1<\lambda<1.2581840884}.\tag{37}$$

Under the CMV candidate classification and the identification of its type-(iv) profile with the regular coordinate model above **[CMV, Lemma 3.8, equation (27), and Proposition 3.9]**, this narrows the unresolved type-(iv) interval from

$$1<\lambda<\frac4\pi$$

to

$$1<\lambda<1.2581840884.$$

The Lean theorem proves this exclusion for the explicit regular coordinate model. The passage to all source-level type-(iv) candidates uses the cited CMV classification and parameterization. The result does not assert existence of a type-(iv) minimizer inside the remaining interval.
